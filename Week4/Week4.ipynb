## Function Declarations!

We can create custom functions that do specific tasks in Python! I'll use the above running time in seconds conversion as an example!

```python
def calculate_seconds (time_string): # def + name of function + arguments in parentheses + colon
    #we are expecting a string formatted like "30:15" as input
    # which will be represented by the placeholder "time_string"
    minutes,seconds =time_string.split(":") #splitting seconds
    total_seconds = int(minutes)*60 + int(seconds) # converting + calculating total seconds
    return total_seconds

print(calculate_seconds("40:45"))
```

Function declarations are made of the following elements:
- the 'def' keyword: tells python that we are defining a new function
- -the name of the new function('calculate_seconds')
- -the arguments: multiple arguments are seperated by commas
- the arguments are then processed by the lines of code that are indicated below the "def" line
- return specifies what the data is output by the function

```python
# another example for generating angry text from normal text
def angry(text):
    text = text.upper() #makes text uppercase
    exclamations = "!"*len(text)
    return text + exclamations

print(angry("hello world"))
```

## Conditionals
conditionals, like for loops, are another fundamental structure found across programming languages

conditionals rely on comparision operators, which compare two values and return a value of true or false
- x == y : are the values equal?
- x != y : are the values unequal?
-  x < y : is the first value less than
-  x > Y : is the first greater than
-  x <= y: less than or equal to
-  x >= y: greater than or equal to
-  x > y and a == b : are both conditions true?
-  x < y or a > b : is at least one of these true?

```python
x = 5
y = 6
a = 1 
b = 1

print(x>y)
print(x<y)
print(x==y)
print(x<y and a==b)
print(x>y or a>b)
```

In python we can run a block of code conditionally that is, only run the code if the condition is true using the if and else syntax:

```python
temperature = 45
if temperature < 60:
    print("put on a jacket")
else:
    print("no jacket needed!")
```

In the above simple example we see the following elements:
-the if keyword, which initializes the consitional statement
-the comparision( temperature < 60) which is evaluated as true or false
-if the comparision evaluates to true, the indented portion directky below the if statement runs
-if the comparison is not ture the indentated portion below else runs

we can also stack conditions using the else if, or elif keyword:

```python
temperature = 30
if temperature >= 80:
    print("wear shorts")
elif temperature >= 65:
    print("wear pants")
elif temperature >= 50:
    print("wear a sweater")
elif temperature >=35:
    print("wear a jacket")
else:
    print("stay home")
```

# working with CSV data
CVS is one of the most common dataset formats, it's basically a spreadsheet with rows and colums
Each row is a new line, each colum is seperated by a comma
CSVs usually have a "header" row that is in the first row of the document
Each subsequent row is actual data

to work with CSV data in python we need to import the CSV library, by typing 'import cvs' before attempting to use any csv functions
we can open a file with the following syntax:
'with open(file_name, options) as file:'

```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file) # initializing an object that interprets CSV data
    header = next(reader) # tells the reader to read the first line and move onto the next
    rows = list(reader) # this takes every other row in the data set and makes it into a list

print(header)
print(rows[0])
```

We can also generate a list of dictionaries that use the header to establish a set of shared keys, assigning values for each row based on those keys and values,

we can do that using the csv method 'DictReader(file)'

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # this will interpret the csv as a list of dictionaries
    data = list(reader)
print(data[-1])
```

```python
#filter by nationality only swedish artist

swedish_artists = []
for artist in data: # for each artist in my data
    if artist["Nationality"] == "Swedish":
        swedish_artists.append(artist)

print(len(swedish_artists))

for artist in swedish_artists[:10]:
    print(artist["DisplayName"])
```
