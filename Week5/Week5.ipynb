first I want to bring in the moma_artist.csv dataset

```python
import csv # brings in all the csv functions 
# the open funtion name/ path
# "r" for read
# and the encoding type
with open("moma_artists.csv","r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # read the file as dictionary data
    data = list(reader) # convert the dictionary object into a usable list called "data"

print(data[0])
```

Simple search of the data by First Name!

```python
name_of_artist = data[0] ['DisplayName']
print(name_of_artist)
```

```python
first_name_of_artist = name_of_artist.split(" ") [0]
print(first_name_of_artist)
```

I want to generate a list of all artists with the first name Mary:

```python
# step 1: make an empty list to hold artists with the name Mary
artists_named_mary = [] # square brackets with no contents means empty list
# step 2: create a for loop that goes through every item in the dataset
for artist in data:
    # estep 3: extract the first name 
    first_name = artist['DisplayName'].split(" ")[0]
    # check if the first name matches "Mary"
    if first_name == "Mary":
        # add our Mary to the new list
        artists_named_mary.append(artist)
print(len(artists_named_mary))
```

Next example shows re-sorting the Moma artist dataset by birth year ('BeginDate')

```python
artists_by_year = sorted(data, key=lambda artist: int(artist['BeginDate']))
# sorted takes two parameters:
# first one is the dataset (here it's called "data", which is the variable containing the complete MoMa set
# second one is the sorting logic, which is represented by the lambda function
# here that lambda function is saying "for each artist in the list, extract the BeginDate and convert it to an int
print(artists_by_year[0])   
```

```python
artist_with_year = []
for artist in data:
    if artist['BeginDate'] != "0": # != means "begin date ISNT zero
        artist_with_year.append(artist)
artists_by_year_filtered = sorted(artist_with_year, key=lambda artist: int(artist['BeginDate']))
print(artists_by_year_filtered[0])
        
```
