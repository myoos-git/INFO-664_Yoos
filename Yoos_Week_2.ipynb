## Converting Data Types
Data can be converted from one type to another using conversion functions:

```python
pi = 3.14159
print(pi)

pi_as_string = str(pi) #str function converts to a string
print(pi_as_string)
print(type(pi_as_string))

pi_as_float = float(pi)
print(pi_as_float)
print(type(pi_as_float))

print(pi_as_float > 3)
```
