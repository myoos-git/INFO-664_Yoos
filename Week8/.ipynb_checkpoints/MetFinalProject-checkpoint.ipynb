## This notebook is specifically to filter the MET Dataset 

```python
import pandas as pd # this imports the pandas library under the nickname 'pd' which is lused a lot of online documentation
df = pd.read_csv('MetObjects.csv',low_memory=False)
df.info()
```

```python
print(df.columns)
```

```python
filtered_df = df.drop('Is Highlight', axis=1)
df.head(3)
```

```python
unwanted_columns = unwanted_columns = [ 'Is Highlight', 'Is Timeline Work', 'Is Public Domain',
       'Gallery Number', 'Period', 'Dynasty',
       'Portfolio', 'Constituent ID', 'Artist Role', 'Artist Prefix',
       'Artist Display Bio', 'Artist Suffix',
       'Artist Alpha Sort', 'Artist Gender', 'Artist ULAN URL',
       'Artist Wikidata URL',  'Dimensions', 'Geography Type', 'City', 'State', 'County',
       'Subregion', 'Locale', 'Locus', 'River', 
       'Rights and Reproduction', 'Object Wikidata URL',
       'Metadata Date', 'Tags AAT URL','Tags Wikidata URL'] 

for column in unwanted_columns:
    df = df.drop(column, axis=1)
```

```python
df.head(6)
```

```python
df.value_counts('Department').plot(kind = 'barh')
```
