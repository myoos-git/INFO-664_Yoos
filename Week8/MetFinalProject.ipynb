## This notebook is specifically to filter the MET Dataset 

```python
import pandas as pd # this imports the pandas library under the nickname 'pd' which is lused a lot of online documentation
df = pd.read_csv('MetObjects.csv',low_memory=False)
df.info()
```

```python
print(df.columns)
```

```python
unwanted_columns = unwanted_columns = [ 'isHighlight', 'isPublicDomain', 'primaryImage', 'additionalImages',
       'constituents','period', 'dynasty', 'reign', 'artistDisplayName', 'artistDisplayBio', 'artistBeginDate', 'artistEndDate', 'artistGender',
       'artistWikidata_URL', 'artistULAN_URL', 'dimensions', 'measurements', 'geographyType', 'city', 'state', 'county', 'tags', 'objectWikidata_URL', 'isTimelineWork',
       'GalleryNumber', 'message'] 

for column in unwanted_columns:
    df = df.drop(column, axis=1)
```
