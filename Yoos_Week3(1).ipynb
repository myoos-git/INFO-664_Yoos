## Parsing comma-delimited data
CSV format is comma seperated values
python has methods for parsing data out of strings by delimiting characters. For instance, commas are oftern used to seperate values in CVS-style datasets. The 'split(",") method will serparate a string based on locations of commas

```python
csv_data = "Nirvana, Smells like teen spirit, 5:01, 1991"
artist, song, length, year = csv_data.split(",")
my_string = f"My favorite band is {artist}, and their best song is {song}, which came out in {year}(), and is {length} long"
print(my_string)
```

## Slicing a string

We can extract a certain range of characters from a string using the slicing approach:

```python
my_text = "saxophone"
print(my_text[:3])
my_text = "saxophone"
print(my_text[4:])
my_text = "saxophone"
print(my_text[2:5]) #shows characters 2-5
print(len(my_text)) #shows length of text
print(my_text[0]) #list starts at 0
my_text = "saxophone"
print(my_text[len(my_text)-1]) #prints the last character by counting from the end
print(my_text[-4]) #this is useful for dates when only looking for the year -4 last 4 characters from the end
```

## All about lists
list is a is all the same thing- type of thing and set of things (value) values are seperated by a comma

```python
birds = ["Robin", "Sparrow", "Blue Jay", "Cardinal", "Crow",
         "Finch", "Owl", "Hawk", "Hummingbird", "Woodpecker"]
leap_years = [2000, 2004, 2008, 2012, 2016, 2020, 2024]
bands = [
    ["John Lennon", "Paul McCartney", "George Harrison", "Ringo Starr"],
    ["Mick Jagger", "Keith Richards", "Charlie Watts", "Ronnie Wood", "Bill Wyman"],
    ["John Phillips", "Michelle Phillips", "Cass Elliot", "Denny Doherty"],
]
print(birds)
print(leap_years)
print(bands)
```

you can access data in a list using its *index number:*


```python
birds = ["Robin", "Sparrow", "Blue Jay", "Cardinal", "Crow",
         "Finch", "Owl", "Hawk", "Hummingbird", "Woodpecker"]
print(birds[0]) #this will give first bird
print(birds[0:3]) # this will print a range of birds (indexing numbers 0-3)
```

if we want to access a value within a list of lists we must use two index numbers:

```python
print(bands[0]) #this gives me a list cntained in the "list of lists"
print(bands[0][-1]) # the -1 index gives me the last entry
print(bands[2][0]) #gives me the first listed meber of band index 2
#list of lists vs nested list
```

## adding items to a list

```python
print(birds)
birds.append('Tucan') #append adds new element to the end of the list
print(birds)
birds.insert(3, 'Cockatoo')
print(birds)
birds.pop(5) #pop removes an element from a list based on index number
print(birds)
birds.remove('Sparrow') #remove removes an element from a list based on value
print(birds)
```

## Reordering a list using sort()

```python
birds.sort() #sort will put our list of birds in alphabetical order!
print(birds)
birds.sort(reverse=True) #this will sort in reverse order
print(birds)
# the same goes for numbers, they will be put into ascending or descending numerical order:
temperatures = [45, 98, 33, 106, 88, 98, 45]
temperatures.sort(reverse=True)
print(temperatures)
```

## Simple statistics on a list: min, max, average

```python
print(temperatures)
print(min(temperatures)) #this gives the minimun value in a list
print(max(temperatures)) # this gives the maximum value

print(sum(temperatures)/len(temperatures)) #this gives the average value of a list
```

## Adding new keys and values to a dictionary
my goal is to look at the "running_time" value , extract a length in seconds by doing the some simple math, and then creating a new key in each respective dictionary to represent the "running_time_in_seconds" as an int

```python
print(beatles_albums[0 ]["running_time"]) #accessing the running_time key in the dictionary
```

```python
minute_string, second_string = beatles_albums [0]["running_time"].split(":")
print(minute_string)
print(second_string)
```

```python
total_seconds = int(minute_string) * 60 + int(second_string)
# converting minute and second strings to ints, multiplying minutes by 60 to get seconds
# and then adding the remaining seconds
print(total_seconds)
```

```python
beatles_albums[0]["running_time_in_seconds"]= total_seconds
print(beatles_albums[0])
```
