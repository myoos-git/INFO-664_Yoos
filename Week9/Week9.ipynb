## Filtering data

we can filter dataframes using comparison operators like ==

comparing a column to a value will return a series of severy row in that column, and whether that matches the criteria 

```python
df['title'] == 'Woman'
```

```python
woman_mask = df['title'] == 'Woman' #store the series in a new variable called woman_mask
type(woman_mask)woman_mask = df['title'] == 'Woman' #store the series in a new variable called woman_mask
type(woman_mask)
```

the above series is called a *mask* which identifies the filtered content as rows flagged True or False. We can use that *mask* to query the dataframe , which will return 

```python
df[woman_mask].info()
```

```python
woman_objects = df[woman_mask]

woman_objects 
```

we can also use df['title'].str.contains(Search_term, na=False

## graphing or plotting data

pandas can generate and charts using '.plot()' . we can generate a variety of types of graphs, including:

-'bar'
-'barh' horizontal bars
-'pie'
-'line'
-'hist'
-'scatter'

bar charts are useful when dealing with the value_counts series for a given column

```python
df.value_counts('department').plot(kind = 'barh')
```

```python
df.value_counts('title').plot(kind = 'bar')
```

```python
df.value_counts('culture').nlargest.plot(kind = 'pie')
```

```python
df.value_counts('artistDisplayName').nlargest(10).plot(kind = 'barh', ylabel= 'artist name', title= 'works with woman in the title')
```
